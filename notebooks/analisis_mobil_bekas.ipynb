# Analisis Data Mobil Bekas

Notebook ini digunakan untuk memahami data, melakukan cleaning, dan menghasilkan insight awal sebelum data dipakai pada dashboard Streamlit. Jalankan dari folder utama proyek.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "data").exists():
    ROOT = ROOT.parent
DATA = ROOT / "data" / "used_cars_clean.csv"
df = pd.read_csv(DATA)
print("Ukuran data bersih:", df.shape)
df.head()
```

## 1. Memahami struktur dan kualitas data

```python
display(df.info())
display(df.describe(include="all").T)
print("Missing values:\n", df.isna().sum())
print("Duplikasi:", df.duplicated().sum())
```

## 2. Ringkasan indikator utama

```python
print("Jumlah listing:", len(df))
print("Rata-rata harga:", round(df.harga.mean(), 2))
print("Median harga:", round(df.harga.median(), 2))
print("Rata-rata jarak tempuh (km):", round(df.jarak_tempuh_km.mean(), 2))
print("Rentang tahun:", int(df.tahun.min()), "-", int(df.tahun.max()))
```

## 3. Distribusi kategori

```python
display(df["kota"].value_counts().head(10).rename("jumlah_listing").to_frame())
display(df["merek"].value_counts().head(10).rename("jumlah_listing").to_frame())
display(df["bahan_bakar"].value_counts().to_frame())
```

## 4. Visualisasi eksploratif

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
df.groupby("tahun")["harga"].mean().plot(ax=axes[0], marker="o", title="Rata-rata harga menurut tahun produksi")
axes[0].set_xlabel("Tahun produksi"); axes[0].set_ylabel("Harga rata-rata (₹)")
df["bahan_bakar"].value_counts().plot(kind="bar", ax=axes[1], title="Jumlah listing menurut bahan bakar")
axes[1].set_xlabel("Jenis bahan bakar"); axes[1].set_ylabel("Jumlah listing")
plt.tight_layout(); plt.show()

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(df["jarak_tempuh_km"], df["harga"], alpha=.5)
ax.set(title="Jarak tempuh vs harga", xlabel="Jarak tempuh (km)", ylabel="Harga (₹)")
plt.show()
```

## 5. Korelasi dan insight

```python
corr = df[["tahun", "harga", "jarak_tempuh_km"]].corr(numeric_only=True)
display(corr)
print("Kota listing terbanyak:", df.kota.value_counts().idxmax())
print("Merek listing terbanyak:", df.merek.value_counts().idxmax())
print("Bahan bakar terbanyak:", df.bahan_bakar.value_counts().idxmax())
print("Korelasi jarak tempuh dan harga:", round(df[["jarak_tempuh_km", "harga"]].corr().iloc[0,1], 3))
```

- Dataset merupakan listing mobil bekas, sehingga harga bukan selalu harga transaksi akhir.
